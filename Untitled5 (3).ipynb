```python
import os
from pathlib import Path
import re
import torch
from datasets import load_dataset, Dataset
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments, DataCollatorForLanguageModeling
print("Imports done. Torch visible:", torch.cuda.is_available())
```

```python
!wget -q -O sample_story.txt "https://www.gutenberg.org/cache/epub/1342/pg1342.txt"  # Pride and Prejudice
txt_files = ["sample_story.txt"]
print("Downloaded and ready:", txt_files)

```

```python
def clean_gutenberg(text):
    # Remove Project Gutenberg header/footer using common markers
    # This is a simple heuristic — you can refine if needed.
    # Remove BOM
    text = text.replace('\ufeff', '')
    # Cut before the start marker
    start_marks = ['*** START OF THIS PROJECT GUTENBERG EBOOK', '***START OF THE PROJECT GUTENBERG EBOOK', 'START OF THE PROJECT GUTENBERG EBOOK']
    end_marks = ['*** END OF THIS PROJECT GUTENBERG EBOOK', '***END OF THE PROJECT GUTENBERG EBOOK', 'END OF THE PROJECT GUTENBERG EBOOK']
    start_idx = 0
    for m in start_marks:
        i = text.find(m)
        if i != -1:
            # try to find newline after marker
            j = text.find('\n', i)
            start_idx = j+1 if j!=-1 else i+len(m)
            break
    end_idx = len(text)
    for m in end_marks:
        i = text.find(m)
        if i != -1:
            end_idx = i
            break
    return text[start_idx:end_idx].strip()

# Read uploaded files and combine
texts = []
for f in txt_files:
    with open(f, 'r', encoding='utf-8', errors='ignore') as fh:
        raw = fh.read()
        cleaned = clean_gutenberg(raw)
        if len(cleaned) > 100:  # skip tiny files
            texts.append(cleaned)

combined_text = "\n\n".join(texts)
print("Combined length (chars):", len(combined_text))
```

```python
# Split combined text into lines or use as single example for block tokenization
# We'll save as one long example, then use dataset.map to create training blocks.
hf_dataset = Dataset.from_dict({"text":[combined_text]})
print(hf_dataset)
```

```python
MODEL_NAME = "distilgpt2"   # or "gpt2"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
# GPT-2 tokenizer sometimes doesn't have pad token; add if missing
if tokenizer.pad_token is None:
    tokenizer.add_special_tokens({'pad_token': '[PAD]'})
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
model.resize_token_embeddings(len(tokenizer))
print("Loaded tokenizer and model:", MODEL_NAME)
```

```python
# Set block size (sequence length). 512 or 1024 depending on GPU memory.
block_size = 512

def tokenize_and_split(example):
    tokens = tokenizer(example["text"], return_attention_mask=False)
    input_ids = tokens["input_ids"]
    # split into blocks of block_size
    chunks = [input_ids[i:i+block_size] for i in range(0, len(input_ids) - block_size + 1, block_size)]
    return {"input_ids": chunks}


tokenized = hf_dataset.map(lambda ex: tokenizer(ex["text"]), batched=True, remove_columns=["text"])

def group_texts(examples):
    # Concatenate all input_ids then chunk
    all_ids = sum(examples["input_ids"], [])
    result = {"input_ids": [all_ids[i:i+block_size] for i in range(0, len(all_ids)-block_size+1, block_size)]}
    return result


all_input_ids = tokenized[0]["input_ids"]
blocks = [all_input_ids[i:i+block_size] for i in range(0, len(all_input_ids)-block_size+1, block_size)]
train_dataset = Dataset.from_dict({"input_ids": blocks})
print("Number of training blocks:", len(train_dataset))
```

```python
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

training_args = TrainingArguments(
    output_dir="gpt2-gutenberg-finetuned",
    overwrite_output_dir=True,
    num_train_epochs=3,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=8,
    learning_rate=5e-5,
    weight_decay=0.01,
    warmup_steps=200,
    logging_steps=50,
    save_total_limit=2,
    fp16=torch.cuda.is_available(),

trainer = Trainer(
    model=model,
    args=training_args,
    data_collator=data_collator,
    train_dataset=train_dataset,
)
```

```python
trainer.train()
trainer.save_model("gpt2-gutenberg-finetuned")
tokenizer.save_pretrained("gpt2-gutenberg-finetuned")
print("Model saved.")
```

```python
from transformers import pipeline
gen_model = AutoModelForCausalLM.from_pretrained("gpt2-gutenberg-finetuned")
gen_tokenizer = AutoTokenizer.from_pretrained("gpt2-gutenberg-finetuned")
generator = pipeline("text-generation", model=gen_model, tokenizer=gen_tokenizer, device=0 if torch.cuda.is_available() else -1)

prompt = "Once upon a time in a small coastal village,"
outputs = generator(prompt, max_length=300, do_sample=True, top_k=50, top_p=0.95, temperature=0.9, num_return_sequences=2)

for i, out in enumerate(outputs):
    print(f"\n--- Generated sample {i+1} ---\n")
    print(out["generated_text"])
```
